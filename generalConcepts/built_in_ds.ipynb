# List

```python
list=[]

list.append(2) #adding in a list
print(list)


list.append(4)
print(list[-1])


# pop removes the last element from the list
num=list.pop()
print(num)
print(list)


# reversing a list
list1=[3,1,5,4,2]
print(list1)

print(list1[::-1])
```

```python
# adding a new list behind another list

list1=['apple','banana']
list2=['cherry','mango']

list2.append(list1)
print(list2)

list3=['guava','pineapple']
list3.extend(list1)

print(list3)
```

# Dictionary


```python
from collections import defaultdict

dict1={}
dict1["a"]=1 #need to assign key value before performing any operatoin i

print(dict1)



dict2=defaultdict(lambda: 1)  #provide a default value for a nonexistent key in the dictionary elimininting the need for checking if the key exisited or not

dict2['a']+=1
print(dict2)

# using list as a default_factory argument 
dict3=defaultdict(list)

for i in range(5):
    dict3[i].append(i)
    
print("Dictionary with values as list:")
print(dict3)
```
